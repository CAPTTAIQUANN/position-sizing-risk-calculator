```python
# ==============================================================================
# UNIVERSAL CRYPTO FUTURES POSITION SIZE CALCULATOR
# ==============================================================================

# 1. Your 3 Inputs (Exact values from your Excel sheet)
entry_price = 0.87718
stop_loss = 0.87678
risk_dollars = 100.0

# 2. Find the price gap between Entry and Stop Loss
# Using abs() removes any minus sign so the formula works for BOTH Longs and Shorts!
stop_distance = abs(entry_price - stop_loss)

# 3. Calculate how many coins/tokens you are buying or shorting
coin_quantity = risk_dollars / stop_distance

# 4. YOUR UNIVERSAL FORMULA: Convert coin quantity into USDT Order Value
order_value = coin_quantity * entry_price

# 5. Bonus: Automatically detect Long vs. Short and calculate 1:3 Take-Profit
if entry_price > stop_loss:
    direction = "LONG"
    take_profit_1_to_3 = entry_price + (stop_distance * 3)
else:
    direction = "SHORT"
    take_profit_1_to_3 = entry_price - (stop_distance * 3)

# 6. Print the clean results
print("=== UNIVERSAL CRYPTO POSITION SIZE CALCULATOR ===")
print(f"Trade Direction:           {direction}")
print(f"Entry Price (E):           {entry_price}")
print(f"Stop Loss (SL):            {stop_loss}")
print(f"Risk ($):                  ${risk_dollars:.2f}")
print(f"Take Profit (1:3 R:R):     {take_profit_1_to_3:.5f}")
print("-------------------------------------------------")
print(f"Position Size (Coins):     {coin_quantity:,.0f} units")
print(f"Order Value (USDT):        ${order_value:,.0f}")
```
